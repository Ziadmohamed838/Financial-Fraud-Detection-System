```python
df_dropped= df.drop(columns=["V13", "V15", "V22", "V23", "V25", "V26"])
```

```python
df_dropped
```

```python
df_dropped.duplicated().sum()
```

```python
df_dropped.drop_duplicates(inplace=True)
```

```python
df_dropped.columns
```

```python
#remove outliers using iqr method
numeric_cols= ['V1', 'V2', 'V3', 'V4', 'V5', 'V6', 'V7', 'V8', 'V9', 'V10',
       'V11', 'V12', 'V14', 'V16', 'V17', 'V18', 'V19', 'V20', 'V21', 'V24',
       'V27', 'V28']
df_iqr= df_dropped.copy()
for col in numeric_cols:
    Q1 = df_dropped[col].quantile(0.25)
    Q3 = df_dropped[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df_iqr[col] = np.where(df_iqr[col] < lower_bound, lower_bound, df_iqr[col])
    df_iqr[col] = np.where(df_iqr[col] > upper_bound, upper_bound, df_iqr[col])
print(f"Original shape: {df_dropped.shape}, After IQR outlier removal: {df_iqr.shape}")
```

```python
df_iqr.head()
```

```python
df_iqr["Class"].value_counts()
```

```python
df_majority= df_iqr[df_iqr.Class==0]
df_minority= df_iqr[df_iqr.Class==1]
```

```python
from sklearn.utils import resample
```

```python
df_majority_new= resample(df_majority, replace=True,n_samples=10000,random_state=42)
```

```python
df_balanced= pd.concat([df_majority_new, df_minority])
```

```python
# fix the imbalance using hybrid approach smote and weight adjustment in training
from imblearn.over_sampling import SMOTE
X = df_balanced.drop('Class', axis=1)
y = df_balanced['Class']
```

```python
X_res, y_res = SMOTE().fit_resample(X, y)
print(y_res.value_counts())
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_res, y_res, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
cols_to_scale = ['Amount'] + [f'V{i}' for i in range(1, 29)]
df[cols_to_scale] = scaler.fit_transform(df[cols_to_scale])

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

log_model = LogisticRegression()
log_model.fit(X_train, y_train)

y_pred_log = log_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred_log)
conf_matrix = confusion_matrix(y_test, y_pred_log)
class_report = classification_report(y_test, y_pred_log)

print("### Logistic Regression ###")
print(f"Accuracy: {accuracy:.2f}")
print("\nConfusion Matrix:\n", conf_matrix)
print("\nClassification Report:\n", class_report)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns


models = {
    "Logistic Regression": LogisticRegression(),
    "Decision Tree": DecisionTreeClassifier(random_state=42, max_depth=6, min_samples_leaf=15),
    "Random Forest": RandomForestClassifier(random_state=42, n_estimators=100),
    "Naive Bayes": GaussianNB(),
    "KNN": KNeighborsClassifier(n_neighbors=10)
}



for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    cm = confusion_matrix(y_test, y_pred)
    
    plt.figure(figsize=(4,3))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False)
    plt.title(f"Confusion Matrix - {name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix


results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    cm = confusion_matrix(y_test, y_pred).ravel()  # [TN, FP, FN, TP]
    
    results.append({
        "Model": name,
        "TN": cm[0],
        "FP": cm[1],
        "FN": cm[2],
        "TP": cm[3]
    })


df_cm = pd.DataFrame(results)


df_melted = df_cm.melt(id_vars="Model", var_name="Metric", value_name="Count")


plt.figure(figsize=(10,6))
sns.barplot(data=df_melted, x="Model", y="Count", hue="Metric",
            palette={"TN": "skyblue", "FP": "orange", "FN": "red", "TP": "green"})

plt.title("Comparison of Confusion Matrix Elements Across Models")
plt.ylabel("Count")
plt.xlabel("Model")
plt.legend(title="Metric")
plt.show()
```
